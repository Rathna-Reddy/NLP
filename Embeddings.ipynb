```python
from transformers import TFBertModel, AutoTokenizer
from scipy.spatial.distance import cosine
```

# Model Setup
Load the pre-trained BERT model and tokenizer. This model will help us extract embeddings for our analysis

```python
# Defining the model name
model_name = "bert-base-cased"

# Loading the pre-trained model and tokenizer
model = TFBertModel.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

# Function Definition: Predict
Define a function that encodes input text into tensors, which are then fed to the model to obtain embeddings

```python
# Defining a function to encode the input text and get model predictions
def predict(text):
    encoded_inputs = tokenizer(text, return_tensors="tf")
    return model(**encoded_inputs)[0]
```

# Defining the sentences
Set up sentences to analyze

```python
#Example 1:
# Defining the sentences
sentence1 = "There was a fly drinking from my soup"
sentence2 = "There was a fly drinking from my soup"
# sentence2 = "There is a fly swimming in my juice"
# sentence2 = "To become a commercial pilot, he had to fly for 1500 hours." #second fly example

# Tokenizing the sentences
tokens1 = tokenizer.tokenize(sentence1)
tokens2 = tokenizer.tokenize(sentence2)
```

# Tokenization and Model Predictions
Tokenize the sentences and obtain predictions (embeddings) from the model.

```python
# Getting model predictions for the sentences
out1 = predict(sentence1)
out2 = predict(sentence2)
```

# Extracting Embeddings
Extract embeddings specifically for the word "fly" from both sentences.

```python
# Extracting embeddings for the word 'fly' in both sentences
emb1 = out1[0:, tokens1.index("fly"), :][0]
emb2 = out2[0:, tokens2.index("fly"), :][0]
```

# Calculating the Cosine Similarity
Calculate the cosine similarity between the embeddings of the word "fly" from both sentences to measure how context affects meaning.

```python
# calculating the cosine similarity between the embeddings
cosine(emb1, emb2)
```

Here since since sentence1 and sentence2 has the same exact text, cosine similarity of the token "fly" is 0.

Next: let's try another example for "fly"

```python
#Example 2: 
# Defining the sentences
sentence1 = "There was a fly drinking from my soup"
sentence2 = "There is a fly swimming in my juice"

# Tokenizing the sentences
tokens1 = tokenizer.tokenize(sentence1)
tokens2 = tokenizer.tokenize(sentence2)
```

```python
# Getting model predictions for the sentences
out1 = predict(sentence1)
out2 = predict(sentence2)
```

```python
# Extracting embeddings for the word 'fly' in both sentences
emb1 = out1[0:, tokens1.index("fly"), :][0]
emb2 = out2[0:, tokens2.index("fly"), :][0]
```

```python
# calculating the cosine similarity between the embeddings
cosine(emb1, emb2)
```

```python
Here since since sentence1 and sentence2 are different, but fly word has mostly same context,
cosine similarity of the token "fly" here is 0.06798797845840454.

Eventhough, here 'fly' has mostly the same context but since they are different sentences,
the same token 'fly' has different embedding, 
How do we know ? Because the cosine similariy is not 0 here, unlike the first example.
Next: let's try one more example for "fly"
```

```python
#Example 3:
# Defining the sentences
sentence1 = "There was a fly drinking from my soup"
sentence2 = "To become a commercial pilot, he had to fly for 1500 hours." #second fly example

# Tokenizing the sentences
tokens1 = tokenizer.tokenize(sentence1)
tokens2 = tokenizer.tokenize(sentence2)
```

```python
# Getting model predictions for the sentences
out1 = predict(sentence1)
out2 = predict(sentence2)
```

```python
# Extracting embeddings for the word 'fly' in both sentences
emb1 = out1[0:, tokens1.index("fly"), :][0]
emb2 = out2[0:, tokens2.index("fly"), :][0]
```

```python
# calculating the cosine similarity between the embeddings
cosine(emb1, emb2)
```

Here since since sentence1 and sentence2 have completely different context,
cosine similarity of the token "fly" here is 0.40477824211120605, which is much higher than the
previous two examples

so, eventhough the token is same i.e 'fly', it's embedding is formed according to the context, which is also called contextual embedding that captures the meaning within the 
context of the input text.

That's why in the third example the cosine similarity value is much higher than the second example.
